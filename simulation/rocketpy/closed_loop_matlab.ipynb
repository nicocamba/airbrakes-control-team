## Airbrakes control, integration with Matlab control algorithm

```python
CONTROL_START_TIME = 10.0
CONTROL_MAX_MACH = 0.95

def matlab_airbrakes_controller(time, sampling_rate, state, state_history, observed_variables, air_brakes):
    global matlab_ctx, controller_initialized
    global t_log, u_log, mach_log, alt_log, vz_log, printed_debug, altitude_ref_mat, velocity_ref_mat
    global last_control_time, last_u_cmd

    # -----------------------------
    # 1) Leer Mach
    # -----------------------------
    
    # -----------------------------
    # 2) Extraer estado para MATLAB
    # IMPORTANTE:
    # Revisar qué índices de state son altitud y velocidad vertical.
    # -----------------------------
    try:
        altitude = float(state[2])   # revisar índice real
        vz = float(state[5])         # revisar índice real
    except Exception as e:
        print(f"[Controller state read error] {e}")
        air_brakes.deployment_level = 0.0
        return time, air_brakes.deployment_level

    x = matlab.double([[altitude], [vz]])

    mach = abs(vz) / speed_of_sound_isa(altitude)

    # -----------------------------
    # 3) Inicializar MATLAB una sola vez
    # cuando entramos en la zona de control
    # -----------------------------
    if (not controller_initialized) and (time >= CONTROL_START_TIME) and (mach < CONTROL_MAX_MACH):
        try:
            matlab_ctx = eng.kalmanCoastingPhaseInit(x, nargout=1)
            altitude_ref_mat = np.array(matlab_ctx['altitude_ref']).flatten()
            velocity_ref_mat = np.array(matlab_ctx['velocity_ref']).flatten()
            controller_initialized = True
            print(f"[MATLAB controller initialized] t={time:.3f}, mach={mach:.3f}")

        except Exception as e:
            print(f"[MATLAB init error] {e}")
            air_brakes.deployment_level = 0.0
            return time, air_brakes.deployment_level
    
    # -----------------------------
    # 4) Si aún no estamos en control, AB cerrados
    # -----------------------------
    if (not controller_initialized) or (matlab_ctx is None):
        air_brakes.deployment_level = 0.0
        return time, air_brakes.deployment_level
    
    # Al llegar al apogeo o empezar a caer: recoger airbrakes
    if vz <= 0:
        air_brakes.deployment_level = 0.0
        return time, air_brakes.deployment_level

    # -----------------------------
    # 5) Llamada al step MATLAB
    # -----------------------------
    
    if last_control_time is not None and abs(time - last_control_time) < 1e-9:
        air_brakes.deployment_level = last_u_cmd
        return observed_variables
    try:
        matlab_out = io.StringIO()
        u_cmd, matlab_ctx = eng.kalmanCoastingPhaseStep(
            float(time),
            x,
            matlab_ctx,
            nargout=2,
            stdout=matlab_out
        )

        
        u_cmd = float(u_cmd)
        u_cmd = max(0.0, min(1.0, u_cmd))

        last_control_time = time
        last_u_cmd = u_cmd

    
    except Exception as e:
        print(f"[MATLAB step error] {e}")
        u_cmd = 0.0

    # -----------------------------
    # 6) Aplicar despliegue
    # -----------------------------
    air_brakes.deployment_level = u_cmd
    t_log.append(time)
    u_log.append(u_cmd)
    mach_log.append(mach)
    alt_log.append(altitude)
    vz_log.append(vz)
    return time, air_brakes.deployment_level

def speed_of_sound_isa(altitude_m):
    gamma = 1.4
    R = 287.05
    T0 = 288.15
    lapse = -0.0065

    h = max(0.0, altitude_m)
    T = T0 + lapse * h
    T = max(T, 216.65)   # evita valores raros
    return (gamma * R * T) ** 0.5
```
